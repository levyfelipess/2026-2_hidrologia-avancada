# *Setup*

```python
%%time
import numpy as np
import pandas as pd

from hid_avan.eda import head_tail_info

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(context='notebook', style='ticks')
plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['mathtext.fontset'] = 'cm'
```

# Analisando as médias históricas de cada mês

## Criando a climatologia mensal

```python
df = pd.read_csv('../../data/processed/pr-et_monthly_inner-joined.csv', parse_dates=['date'])

df_only_month = pd.concat((df.iloc[:, 0].dt.month, df.iloc[:, 1:]), axis=1)
df_only_month.rename({'date' : 'month'}, axis='columns', inplace=True)
del df

head_tail_info(df_only_month, n=5)
```

```python
df_climatology_pr_et = df_only_month.groupby(
    by='month',
    as_index=False
).agg(
    historical_avg_pr=('agg_total__avg_pr', 'mean'),
    historial_avg_et=('et', 'mean')
)

df_climatology_pr_et.month = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
df_climatology_pr_et.to_csv('../../reports/tables/climatology_pr-et.csv')
head_tail_info(df_climatology_pr_et, 6)
```

## Analisando graficando a climatologia mensal

```python
df_climatology_pr_et = pd.read_csv('../../reports/tables/climatology_pr-et.csv', index_col=0)
head_tail_info(df_climatology_pr_et, 6)
```

```python
fig, ax_pr = plt.subplots(1, 1, figsize=(8, 4), layout='constrained')
ax_etp = ax_pr.twinx()

sns.barplot(ax=ax_pr, data=df_climatology_pr_et, x='month', y='historical_avg_pr', width=.7, label='Precipitações Mensais Médias')
ax_pr.set_ylabel('Pr (mm)')
ax_pr.set_xlabel('Mês')
ax_pr.set_ylim(0, None)
ax_pr.set_title('Climatologia Mensal de Precipitações (Pr) e Evapotranspirações Potenciais (EtP) Médias')

sns.lineplot(ax=ax_etp, data=df_climatology_pr_et, x='month', y='historial_avg_et', color='red', marker='o', label='Evapotranspirações Mensais Médias')
ax_etp.set_ylabel('EtP (mm)')
ax_etp.set_ylim(0, None)

bar, bar_label = ax_pr.get_legend_handles_labels()
line, line_label = ax_etp.get_legend_handles_labels()
ax_pr.legend(bar + line, bar_label + line_label, loc='center right')
ax_etp.get_legend().set_visible(False)

ax_pr.grid(lw=.5, axis='y')

#fig.savefig('../../reports/figures/climatology_pr-et.png', dpi=300)
plt.show()
```
